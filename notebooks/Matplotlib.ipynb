```python
x = [1, 2, 3, 4, 5, 6, 7]
y1 = [20, 15, 7, 21, 12, 14, 22]
y2 = [23, 17, 10, 12, 10, 17, 21.5]
```

```python
plt.plot(x, y1, color="brown", linewidth=1.5, marker="o") #o, O, ^, x, X, s
plt.plot(x, y2, color="violet", linewidth=2.5, marker="X") #o, O, ^, x, X, s
```

```python
plt.plot(x, y1, color="brown", linewidth=1.5, marker="o") #o, O, ^, x, X, s
plt.plot(x, y2, color="violet", linewidth=2.2, marker="x") #o, O, ^, x, X, s
plt.xlabel("Days")
plt.ylabel("Sales")
plt.title("Weekly Performance")
plt.grid()
plt.show()
```

```python
plt.plot(x, y1, label="Bikes", color="Brown", linewidth=1.5, marker="o") #o, O, ^, x, X, s
plt.plot(x, y2, label="Cars", color="violet", linewidth=2.2, marker="x") #o, O, ^, x, X, s
plt.xlabel("Days")
plt.ylabel("Sales")
plt.title("Weekly Performance")
plt.legend(loc="best") #upper left, upper right, lower left, lower right
plt.grid()
plt.show()
```

```python
plt.plot(x, y1, label="Bikes", color="Brown", linewidth=1.5, marker="o") #o, O, ^, x, X, s
plt.plot(x, y2, label="Cars", color="violet", linewidth=2.2, marker="x")
plt.xlabel("Days")
plt.ylabel("Sales")
plt.title("Weekly Performance")
plt.legend(loc="upper left") #upper left, upper right, lower left, lower right, best
plt.grid()
plt.show()
```

```python
plt.figure(figsize=(8, 4)) #(width, height)
plt.plot(x, y1, label="Bikes", color="Brown", linewidth=1.5, marker="o") #o, O, ^, x, X, s
plt.plot(x, y2, label="Cars", color="Violet", linewidth=2.2, marker="x") #o, O, ^, x, X, s
plt.xlabel("Days")
plt.ylabel("Sales")
plt.title("Weekly Performance")
plt.legend(loc="upper right") #upper left, upper right, lower left, lower right, best
plt.grid()
plt.show()
```

```python
plt.figure(figsize=(8, 4), dpi=100) #(width, height)
plt.plot(x, y1, label="Bikes", color="Brown", linewidth=1.5, marker="o")
plt.plot(x, y2, label="Cars", color="Violet", linewidth=2.2, marker="x")
plt.xlabel("Days")
plt.ylabel("Sales")
plt.title("Weekly Performance")
plt.legend(loc="lower left") #upper left, upper right, lower left, lower right, best
plt.grid()
plt.show()
```

```python
plt.figure(figsize=(8, 4), dpi=1000) #(width, height) dpi is increased to 1000
plt.plot(x, y1, label="Bikes", color="Brown", linewidth=1.5, marker="o")
plt.plot(x, y2, label="Cars", color="Violet", linewidth=2.2, marker="x")
plt.xlabel("Days")
plt.ylabel("Sales")
plt.title("Weekly Performance")
plt.legend(loc="lower left") #upper left, upper right, lower left, lower right, best
plt.grid()
plt.show()
```

```python
plt.figure(figsize=(8, 4), dpi=100) #(width, height)
plt.plot(x, y1, label="Bikes", color="Brown", linewidth=1.5, marker="o")
plt.plot(x, y2, label="Cars", color="Violet", linewidth=2.2, marker="x")
plt.xlabel("Days - Very long x axis label to test plt tight layout")
plt.ylabel("Sales - Very long y axis label to test plt tight layout")
plt.title("Weekly Performance")
plt.legend(loc="lower right") #upper left, upper right, lower left, lower right, best
plt.grid()
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 4), dpi=200)

plt.subplot(2, 1, 1) #(Rows, Columns, Position)
plt.plot(x, y1, color="Brown", linewidth=1.5, marker="o")
plt.title("Bike Sales")
plt.ylabel("Sales")
plt.grid()

plt.subplot(2, 1, 2) #(Rows, Columns, Position)
plt.plot(x, y2, color="Violet", linewidth=2.2, marker="x")
plt.title("Car Sales")
plt.xlabel("Days")
plt.ylabel("Sales")
plt.grid()

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 4), dpi=200)
plt.scatter(x, y1, label="Bikes", color="Brown", marker="o")
plt.scatter(x, y2, label="Cars", color="Violet", marker="x")
plt.xlabel("Days")
plt.ylabel("Sales")
plt.title("Weekly Performance")
plt.legend(loc="best") #upper left, upper right, lower left, lower right, best
plt.grid()
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 4), dpi=200)
plt.bar(x, y1, color="Violet", width=0.6)
plt.xlabel("Days")
plt.ylabel("Sales")
plt.title("Weekly Performance")
plt.grid(axis="y")
plt.show()
```

```python
plt.figure(figsize=(8, 4), dpi=150)
plt.barh(x, y2, color="brown")
plt.xlabel("Sales")
plt.ylabel("Days")
plt.title("Weekly Performance")
plt.grid(axis="x")
plt.show()
```

```python
import numpy as np
plt.figure(figsize=(8,4), dpi=200)
x1 = np.arange(len(x))
width = 0.35
plt.bar(x1 - width/2, y1, width, label='Bikes')
plt.bar(x1 + width/2, y2, width, label='Cars')

plt.xticks(x1, x)
plt.xlabel("Days")
plt.ylabel("Sales")
plt.title("Weekly Sales Performance")
plt.legend(loc="best")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,4), dpi=200)
width_of_each_bar = 0.25
x1 = np.arange(len(x))
plt.bar(x1-width_of_each_bar/2, y1, width_of_each_bar, label="Bikes") #Center of the bar
plt.bar(x1-width_of_each_bar/2, y2, width_of_each_bar, label="Cars")
plt.bar(x1+width_of_each_bar/2, y3, width_of_each_bar, label="Scooters")
plt.xlabel("Days")
plt.ylabel("Sales")
plt.title("Weekly Sales Performance")
plt.tight_layout()
plt.show()
```
